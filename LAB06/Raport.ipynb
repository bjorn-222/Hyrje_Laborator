## Qëllimi

Qëllimi i këtij laboratori është të kuptohet matja e madhësive fizike dhe llogaritja e mesatares dhe pasigurisë së matjeve.

## Teoria

Çdo matje përmban pasiguri. Rezultati i një matjeje jepet si:

x = vlera ± pasiguria

Pasiguria mund të vijë nga:
- instrumenti matës
- gabimet njerëzore
- kushtet e matjes

## Metoda

U kryen disa matje për një madhësi dhe u llogarit mesatarja e tyre duke përdorur Python.

```python
import numpy as np

# matje gjatesie (cm)
length = np.array([10.1, 10.2, 9.9, 10.0, 10.1])

# matje mase (g)
mass = np.array([50.2, 50.1, 50.3, 50.2, 50.1])

# matje kohe (s)
time = np.array([2.1, 2.0, 2.2, 2.1, 2.0])

print("Gjatësia:", length)
print("Masa:", mass)
print("Koha:", time)
```

```python
print("Mesatarja e gjatësisë:", np.mean(length))
print("Mesatarja e masës:", np.mean(mass))
print("Mesatarja e kohës:", np.mean(time))
```

```python
print("Devijimi standard (gjatësia):", np.std(length))
print("Devijimi standard (masa):", np.std(mass))
print("Devijimi standard (koha):", np.std(time))
```

## Rezultatet

U llogarit mesatarja dhe devijimi standard për secilën madhësi të matur.

## Përfundimi

Rezultatet tregojnë se matjet kanë variacione të vogla dhe mesatarja përfaqëson vlerën më të besueshme të madhësisë.

## Qëllimi

Qëllimi i këtij laboratori është të përdoret Python për analiza të thjeshta shkencore duke përdorur NumPy dhe Matplotlib.

## Teoria

Python përdoret gjerësisht në analiza shkencore për shkak të bibliotekave të fuqishme si:

- NumPy për operacione numerike
- Matplotlib për grafikë
- SciPy për analiza më të avancuara

## Metoda

Në këtë laborator u krijuan të dhëna numerike dhe u analizuan duke përdorur Python. U ndërtuan grafikë për të paraqitur rezultatet.

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = x**2

print("X:", x)
print("Y:", y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Grafiku i të dhënave")
plt.grid()
plt.show()
```

```python
# mesatarja
print("Mesatarja e Y:", np.mean(y))
```

## Rezultatet

Nga analiza u krijua një grafik që tregon marrëdhënien midis variablave X dhe Y.

## Qëllimi

Qëllimi i këtij laboratori është të llogaritet dendësia e një objekti dhe të kuptohet ndikimi i pasigurive në rezultat.

## Teoria

Dendësia jepet nga formula:

$$
\rho = \frac{m}{V}
$$

ku:
- m është masa
- V është vëllimi

Çdo matje përmban pasiguri që ndikon në rezultatin final.

## Metoda

U përdorën matje të masës dhe vëllimit për të llogaritur dendësinë e një objekti dhe për të analizuar ndryshimet në rezultat.

```python
import numpy as np

# masa (g)
mass = np.array([100.2, 100.1, 100.3])

# vellimi (cm^3)
volume = np.array([50.0, 49.8, 50.2])

print("Masa:", mass)
print("Vëllimi:", volume)
```

```python
density = mass / volume

print("Dendësia:", density)
print("Mesatarja e dendësisë:", np.mean(density))
```

```python
print("Devijimi standard:", np.std(density))
```

```python
import matplotlib.pyplot as plt

plt.plot(density, 'o-')
plt.title("Dendësia për matje të ndryshme")
plt.xlabel("Matja")
plt.ylabel("Dendësia")
plt.grid()
plt.show()
```

## Rezultatet

U llogarit dendësia për çdo matje dhe mesatarja e saj. U vu re një variacion i vogël midis vlerave.

## Përfundimi

Rezultatet tregojnë se dendësia mund të llogaritet nga matjet e masës dhe vëllimit, ndërsa pasiguria ndikon në saktësinë e rezultatit.

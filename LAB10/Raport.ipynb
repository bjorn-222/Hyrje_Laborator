## Qëllimi

Qëllimi i këtij laboratori është të studiohet variacioni i kohës së ekzekutimit duke përsëritur matjet dhe analizuar rezultatet statistikisht.

## Teoria

Koha e ekzekutimit nuk është konstante, por ndryshon për shkak të faktorëve si:
- CPU
- memoria
- sistemi operativ

Prandaj ajo trajtohet si një madhësi statistikore.

## Metoda

U përdor një algoritëm i thjeshtë dhe u mat koha e ekzekutimit shumë herë për të analizuar shpërndarjen e rezultateve.

```python
import time
import numpy as np
import matplotlib.pyplot as plt
```

```python
def test_algorithm(n):
    total = 0
    for i in range(n):
        total += i
    return total
```

```python
times = []

n = 10000  # input fixed

for _ in range(100):  # repeat 100 times
    start = time.perf_counter()
    
    test_algorithm(n)
    
    end = time.perf_counter()
    times.append(end - start)

print(times[:10])  # print some values
```

```python
mean = np.mean(times)
std = np.std(times)

print("Mesatarja:", mean)
print("Devijimi standard:", std)
```

```python
plt.hist(times, bins=10)
plt.title("Histogrami i kohës së ekzekutimit")
plt.xlabel("Koha (s)")
plt.ylabel("Frekuenca")
plt.grid()
plt.show()
```

## Rezultatet

U vu re se koha e ekzekutimit ndryshon nga një matje në tjetrën dhe formon një shpërndarje rreth një vlere mesatare.

## Diskutim

Variacioni vjen nga faktorë të jashtëm si sistemi operativ dhe përdorimi i CPU. Rezultatet tregojnë se performanca nuk është fikse.

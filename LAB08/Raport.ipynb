## Qëllimi

Qëllimi i këtij laboratori është të analizohet koha e reagimit si një madhësi statistikore dhe të studiohet shpërndarja e saj.

## Teoria

Koha e reagimit nuk është një vlerë e vetme, por një shpërndarje vlerash.

Për analizë përdoren:
- Mesatarja (mean)
- Devijimi standard

Shpesh këto të dhëna ndjekin një shpërndarje normale (Gaussiane).

## Metoda

U simulua një seri matjesh për kohën e reagimit dhe u analizuan statistikisht duke përdorur Python.

```python
import numpy as np

# simulim i kohes se reagimit (sekonda)
reaction_time = np.array([
0.21, 0.19, 0.22, 0.20, 0.23,
0.18, 0.21, 0.20, 0.19, 0.22,
0.21, 0.20, 0.23, 0.19, 0.21
])

print(reaction_time)
```

```python
mean = np.mean(reaction_time)
std = np.std(reaction_time)

print("Mesatarja:", mean)
print("Devijimi standard:", std)
```

```python
import matplotlib.pyplot as plt

plt.hist(reaction_time, bins=5)
plt.title("Histogrami i kohës së reagimit")
plt.xlabel("Koha (s)")
plt.ylabel("Frekuenca")
plt.grid()
plt.show()
```

## Rezultatet

U ndërtua histogrami i të dhënave dhe u llogarit mesatarja dhe devijimi standard i kohës së reagimit.

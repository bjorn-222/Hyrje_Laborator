## Qëllimi

Qëllimi i këtij laboratori është të matet koha e ekzekutimit të algoritmeve dhe të analizohet si ndryshon ajo në varësi të madhësisë së të dhënave.

## Teoria

Koha e ekzekutimit tregon sa kohë i duhet një programi për të përfunduar.

Ajo matet zakonisht në sekonda dhe varet nga:
- madhësia e input-it
- kompleksiteti i algoritmit

## Metoda

U përdor Python për të matur kohën e ekzekutimit të një algoritmi të thjeshtë për madhësi të ndryshme të të dhënave.

```python
import time
import numpy as np
import matplotlib.pyplot as plt
```

```python
def test_algorithm(n):
    total = 0
    for i in range(n):
        total += i
    return total
```

```python
sizes = [1000, 5000, 10000, 20000]
times = []

for n in sizes:
    start = time.time()
    
    test_algorithm(n)
    
    end = time.time()
    times.append(end - start)

print("Sizes:", sizes)
print("Times:", times)
```

```python
plt.plot(sizes, times, 'o-')
plt.xlabel("Madhësia e input-it")
plt.ylabel("Koha (s)")
plt.title("Koha e ekzekutimit")
plt.grid()
plt.show()
```

## Rezultatet

U vu re se koha e ekzekutimit rritet me rritjen e madhësisë së të dhënave.

## Diskutim

Rezultatet tregojnë se algoritmi kërkon më shumë kohë për input më të madh. Kjo tregon lidhjen midis madhësisë së të dhënave dhe performancës.

## Qëllimi

Qëllimi i këtij laboratori është të analizohen të dhënat GPS dhe të vizualizohen ato duke përdorur Python.

## Teoria

Të dhënat GPS përbëhen nga koordinata gjeografike:
- gjerësia (latitude)
- gjatësia (longitude)

Këto të dhëna mund të përdoren për të analizuar lëvizjen dhe pozicionin.

## Metoda

U përdor një dataset i thjeshtë me koordinata dhe u vizualizua trajektorja duke përdorur Python.

```python
import numpy as np

# koordinata (lat, lon)
lat = np.array([41.33, 41.331, 41.332, 41.333, 41.334])
lon = np.array([19.82, 19.821, 19.822, 19.823, 19.824])

print("Latitude:", lat)
print("Longitude:", lon)
```

```python
import matplotlib.pyplot as plt

plt.plot(lon, lat, 'o-')
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Trajektorja GPS")
plt.grid()
plt.show()
```

```python
# llogaritje e thjeshte (jo reale GPS, por ok per lab)
distances = np.sqrt((lat[1:] - lat[:-1])**2 + (lon[1:] - lon[:-1])**2)

print("Distancat midis pikave:", distances)
print("Distanca totale:", np.sum(distances))
```

## Rezultatet

U vizualizua trajektorja e lëvizjes dhe u llogarit distanca midis pikave.

## Diskutim

Rezultatet tregojnë se të dhënat GPS mund të përdoren për të analizuar lëvizjen dhe për të llogaritur distanca.
